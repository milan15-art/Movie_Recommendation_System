```python
                # MOVIE RECOMMENDATION SYSTEM

movies = {
    "Action" :["Bahubali","KGF","Leo","Brat","Marco","Salar"],
    "Comedy" :["3 Idiots","Hera Pheri","Golmaal","chhichhore"],
    "Romance"  :["365 days","Titanic","DDLJ","Notebook"],
    "Horror" : ["Conjuring 2","Stree","1921","Shapit","Raaz"],
    "Sci-Fi" :["Avengers","Intersteller","Matrix","Gravity"]
    
    
}

def show_movie_types():
    print("Available Movies Category ")
    for i,type in enumerate(movies.keys(), start=1):
        print(f"{i}.{type}")
        
def recommend_types(choice):
    categories = list(movies.keys())
    
    
    if 1 <= choice <= len(categories):
        category = categories[choice - 1]
        print(f" Recommended movie type {category}")
        for categor in enumerate(movies[category], start=1):
            print(f"{categor}")
    else:
        print("Invalid value  ")
        
    
        
    
    
    



while True:
    print("1. Show Movies type ")
    print("2. Recommend Category")
    print("3. Exit")
    
    option = int(input("Enter number from (1-3)"))
    
    
    if (option == 1):
        show_movie_types()
    
    elif(option == 2):
        try :
            user_choice = int(input("select type from recommend movies"))
            recommend_types(user_choice)
            
        except  ValueError:
            print("Please enter valid value")
            
            
    elif(option == 3):
        print("Program Ended !")
        break
    else:
        print("Enter valid input")
        
        
        
```
